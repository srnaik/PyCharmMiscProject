```python
import numpy as np
```

```python
first_array = np.arange(1,13)
first_array
```

```python
second_array = np.reshape(first_array,(3,4))
second_array
```

```python
third_array = np.reshape(first_array,(6,2))
third_array
```

```python
fourth_array = np.reshape(first_array,(3,2,2))
fourth_array
```

```python
fifth_array = np.array([[1,2],[3,4],[5,6]])
fifth_array
```

```python
sixth_array = np.reshape(fifth_array,-1)
sixth_array
```

```python
seventh_array = fifth_array.flatten()
eighth_array = fifth_array.ravel()
eighth_array[0] = 8
print("Seventh Array: ", seventh_array)
print("Eighth Array: ", eighth_array)
print("Fifth Array: ", fifth_array)
```
